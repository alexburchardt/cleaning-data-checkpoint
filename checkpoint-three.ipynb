## Missing Data

```python
na_count=df.isnull().sum()
columns_with_na=na_count[na_count>0]
print(columns_with_na)
#I found n/a values in certain columns. There are a significant number in each of those columns, however, so I looked closer at the data. There are legitimate reasons for those columns to have n/a values, so it isn't necessarily an error for them to exist.
#For example, the n/a values in the "Data_Value" column are because those rows represent where the sample size was too small for that particular demographic or the relative standard error was too large. Instead of including bad data or making it appear
#that it wasn't collected, the researchers included the n/a values with an explanation in the "Data_Value Footnote" column.
```

## Unnecessary Data

Look for the different types of unnecessary data in your dataset and address it as needed. Make sure to use code comments to illustrate your thought process.

```python

#There are some columns I know I can drop because they are repetitive. For example, the "Datasource" column is the same for every single row. 
#"LocationDesc" is the same information as "LocationAbbr". "DataValueTypeID" and "Data_Value_Unit" are both the same as "Data_Value_Type".
#"Data_Value_Alt" is the same information as "Data_Value" so I'm also dropping one.
df=df.drop(columns=["Datasource","LocationDesc", "DataValueTypeID", "Data_Value_Unit", "Data_Value_Alt", "Geolocation", "LocationID", "StratificationCategoryID1"])

#Next, I know I'm only interested in looking at Missouri and the US as a whole. I'm going to remove rows for other locations.
df=df[df['LocationAbbr'].isin(['US', 'MO'])]
```
